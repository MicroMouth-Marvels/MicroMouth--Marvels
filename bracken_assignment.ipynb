# **Metagenomic Profile Analysis Using Bracken.**

Metagenomics is a culture-independent tool that gives insight into which organisms are present and what they are doing through the study of their genes in an environmental sample. The goals of metagenomics is to find new genes with desired biological activity and study environmental microbes without culturing them. Metagenomics uses DNA sequencing techniques to study DNA extracted directly from environmental samples.

A Bracken (Bayesian Reestimation of Abundance with KrakEN) report typically provides information about the relative abundance of different taxa in a metagenomic sample. Bracken usually uses the output report generated from the kraken2 software program with the classified read counts and estimates the abundance of each taxon in samples. The report finally generated by bracken is visualized using some python libraries and pavian. 

The relative abundance is crucial for understanding the composition of the sample. The report helps in identifying which organisms are present in the sample and in what proportion. 

In this analysis, We will be identifying the organisms present in this sample and in which proportion by visualizing them using python libraries. Track the species abundance over the six different time points (t0 to t5). Also detect any form of dysbiosis that can impact the health of the gut microbiome. The organisms present is a combination of plant species and organisms found in the human gut.

# **Opening, Loading and Reading data...**

In a typical bracken report generated from kraken has the following columns as shown in the pandas dataframe below:

1. Name of the species : The taxonomic name (e.g., Bacteria, Escherichia coli).
2. Taxonomy_id: The NCBI taxonomy ID for the taxon.
3. Taxonomy_lvl: The taxonomic level (e.g., D for Domain, P for Phylum, G for Genus, S for Species).
4. Kraken_assigned_reads: The number of reads originally assigned to this taxon by Kraken.
5. Added_reads: The number of reads that Bracken added to this taxon after re-estimation.
6. New_est_reads: The total number of reads estimated for this taxon after Bracken’s re-estimation.
6. Fraction_total_reads: The relative abundance of this taxon, calculated as the fraction of total reads assigned to this taxon. It represents the relative abundance of different taxa in a sample, expressed as a fraction of the total number of reads. This column should ideally sum up to 1

```python
# Loading the file
file_path = r"C:\Users\Snowwhite Lily\Downloads\project codes\micromouth_bracken.species.xlsx"
sheet_names = ['t0', 't1', 't2', 't3', 't4', 't5']

# Reading data by specifying the engine explicitly
data = {}
for sheet in sheet_names:
    try:
        # Attempting to read the Excel file using the 'openpyxl' engine
        data[sheet] = pd.read_excel(file_path, sheet_name=sheet, engine='openpyxl')
    except ValueError:
        # Attempting to read the Excel file using the 'xlrd' engine
        data[sheet] = pd.read_excel(file_path, sheet_name=sheet, engine='xlrd')

# Combining data into a single DataFrame with a 'Timepoint' column
combined_data = pd.DataFrame()
for sheet in sheet_names:
    df = data[sheet]
    df['Timepoint'] = sheet
    combined_data = pd.concat([combined_data, df])

# Now visualize the data
plt.figure(figsize=(12, 8))

# Make a line plot of abundance over time for each species with a distinct color palette
sns.lineplot(data=combined_data, x='Timepoint', y='fraction_total_reads', hue='name', marker='o', palette='tab20')

plt.title('Species Abundance Over Time')
plt.xlabel('Timepoint')
plt.ylabel('Abundance')
plt.legend(title='Species', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

# Save or show the plot
plt.savefig('species_abundance_over_time.png')
plt.show()
```

The human gut microbiome begins to form at birth and evolves over a lifetime, becoming home to trillions of microorganisms, predominantly bacteria. Disruptions in this microbial balance, known as dysbiosis, have been linked to various diseases, including cancer, cardiovascular disease, inflammatory bowel disease, and neurological disorders. Researchers are investigating specific genera and species within the microbiome to understand their collective impact and identify individual microbes that may influence disease outcomes.
